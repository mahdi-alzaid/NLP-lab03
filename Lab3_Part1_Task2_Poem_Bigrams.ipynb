## N-gram Generation from Poem Data

**Goal:**
1. Load the Poem Classification (NLP) dataset
2. Pre-process the text (tokenize, remove punctuation, lowercase, remove stopwords)
3. Generate bi-grams (2-grams) from the preprocessed text
4. Calculate the frequency of each unique bi-gram
5. Display the top 10 most common bi-grams

**Dataset:** `Poem_classification - train_data.csv` (Kaggle: Poem Classification (NLP))

## 1. Import Libraries

```python
import pandas as pd
import string
from collections import Counter

import nltk
from nltk.corpus import stopwords
from nltk import bigrams
# nltk.download('stopwords')   # uncomment on first run
```

## 2. Load the Dataset

```python
df = pd.read_csv('Poem_classification - train_data.csv')
print('Shape:', df.shape)
df.head()
```

```python
# A few rows have no poem text (NaN) - drop them before processing
df = df.dropna(subset=['Poem']).reset_index(drop=True)
print('Rows after dropping empty poems:', df.shape)
```

## 3. Pre-process the Text

- Lowercase
- Remove punctuation
- Tokenize
- Remove English stopwords, keep only alphabetic tokens

```python
stop_words = set(stopwords.words('english'))
punct_table = str.maketrans('', '', string.punctuation)

def preprocess(text):
    text = str(text).lower()
    text = text.translate(punct_table)
    tokens = text.split()
    tokens = [w for w in tokens if w.isalpha() and w not in stop_words]
    return tokens

df['tokens'] = df['Poem'].apply(preprocess)
df[['Poem', 'tokens']].head()
```

## 4. Generate Bi-grams

```python
df['bigrams'] = df['tokens'].apply(lambda toks: list(bigrams(toks)))
df[['tokens', 'bigrams']].head()
```

## 5. Calculate Bi-gram Frequencies

```python
all_bigrams = [bg for row in df['bigrams'] for bg in row]
bigram_freq = Counter(all_bigrams)

print('Total bigrams generated:', len(all_bigrams))
print('Unique bigrams:', len(bigram_freq))
```

## 6. Top 10 Most Common Bi-grams

```python
top10 = bigram_freq.most_common(10)

top10_df = pd.DataFrame(
    [(f'{w1} {w2}', count) for (w1, w2), count in top10],
    columns=['Bigram', 'Count']
)
top10_df
```

```python
print('Top 10 most common bi-grams in the poem dataset:')
for (w1, w2), count in top10:
    print(f'  ({w1}, {w2}): {count}')
```
